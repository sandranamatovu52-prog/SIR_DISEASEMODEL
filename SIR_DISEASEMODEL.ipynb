```python
import numpy as np
import matplotlib.pyplot as plt

# 1. MODELING QUESTION: How will disease spread in a small population of 100 people?

# 2. STATE VARIABLES AND UNITS
# S = Susceptible people [persons]
# I = Infected people [persons] 
# R = Recovered people [persons]
# t = Time [days]

# 3. ASSUMPTIONS:
# a) Population is closed (no birth/death/migration)
# b) Everyone mixes equally
# c) Recovered people get immunity

# 4. INPUTS, PARAMETERS, OUTPUTS
# Inputs: Initial values S0, I0, R0
# Parameters: beta=transmission rate, gamma=recovery rate, N=total population
# Outputs: S, I, R over time

N = 100
S0 = 90
I0 = 10
R0 = 0
beta = 0.3   # infection rate
gamma = 0.1  # recovery rate
t_max = 100  # days

# Initialize
S = np.zeros(t_max+1)
I = np.zeros(t_max+1)
R = np.zeros(t_max+1)
S[0] = S0
I[0] = I0
R[0] = R0

# 5 & 6. FLOW DIAGRAM: S -> I -> R
# 7. UPDATE RULE
for t in range(t_max):
    S[t+1] = S[t] - beta * S[t] * I[t] / N
    I[t+1] = I[t] + beta * S[t] * I[t] / N - gamma * I[t]
    R[t+1] = R[t] + gamma * I[t]

# Plot
plt.figure(figsize=(10,6))
plt.plot(S, label='Susceptible')
plt.plot(I, label='Infected')
plt.plot(R, label='Recovered')
plt.xlabel('Time (days)')
plt.ylabel('People')
plt.title('SIR Disease Spread Model')
plt.legend()
plt.grid()
plt.show
plt.savefig('SIR_graph.png')
```
